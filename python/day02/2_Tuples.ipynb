```python
"""
==========================================================
DAY 2 - PYTHON TUPLES
==========================================================

A tuple:
- Is ordered
- Allows duplicates
- Is immutable
- Uses ()

Example:
    numbers = (10, 20, 30)
==========================================================
"""
```

```python
# creating tuple 

n = (1,2,3,4,5,6,7,8,1)
```

```python
# Access Element

print(n[1])
print(n[6])
```

```python
# tupple unpacking 

student = ("Rishabh",10,"Mumbai")

name,age,place = student

print(name)
print(age)
print(place)
```
