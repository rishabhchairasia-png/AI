```python
"""
==========================================================
DAY 2 - PYTHON SETS
==========================================================

A set:
- Stores unique values
- Does not maintain normal list-style indexing
- Is mutable
- Uses {}

Main use:
    Removing duplicates
==========================================================
"""
```

```python
# Creating a set

n = { 1,2,3,4,4,5,9,0}
```

```python
# Automatic dulicate delete

print(n)
```

```python
# ADD

n.add(90)

print(n)
```

```python
# all unique value 

x = {9,8,6,56,87,87}

print( n | x)
```

```python
# Common value only 

print(n & x)
```

```python
# Intersection 

print(n - x)
```

```python
# Remove dulicate from list 

a = [1,1,2,12,3,4,5,3,2,1,1]

u_a = list(set(a))
print(u_a)
```
