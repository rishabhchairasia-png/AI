```python
"""
==========================================================
DAY 2 - PYTHON LISTS
==========================================================

A list:
- Stores multiple values
- Is ordered
- Allows duplicate values
- Is mutable (can be changed)

Syntax:
    list_name = [value1, value2, value3]

Example:
    numbers = [10, 20, 30]
==========================================================
"""
```

```python
# Creating a list

n = [1,2,3,6,7,2,8,9,6,8,9,3,4,5,6,7,8,9]
print(n)
```

```python
# Accessing elements of a list

print(n[0])
print(n[6])
print(n[1])
```

```python
# Negtaive indexing in list

print(n[-1])
print(n[-2])
```

```python
# Slicing in list

print(n[2:8])
```

```python
# Appending elements to a list

n.append(10)
print(n)
```

```python
# Inserting elements to a list

n.insert(0,0)

print(n)
```

```python
# Removing elements from a list

n.remove(1)

print(n)
```

```python
# popping elements from a list

n.pop(0)
print(n)
```

```python
# sort

n.sort()

print(n)
```

```python
# Loop through a list

for i in n:
    print(i , end = " ")
```

```python
if 20 in n:
    print("20 exist")
else:
    print("Not exist")
```
