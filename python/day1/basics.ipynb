```python
# Variables & Data type 

name = "Rishabh"
age = 19
grade = 8.5
is_student = True

print(type(name))  # Output: <class 'str'>
print(type(age))   # Output: <class 'int'>
print(type(grade)) # Output: <class 'float'>
print(type(is_student))  # Output: <class 'bool'>
```

```python
# Type conversion 

age = "19"
age = int(age)

print(type(age))

x = "10"
y = "10"

print(int(x) + int(y))
```

```python
# input and output

name = input("Enter your name :")
age = int(input("Enter your age :"))

print("Hello " + name )
print("You are", age , "old")
```

```python
# Operations 

a = 10 
b = 3

print(a + b )
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

```python
# Comparison operators 

a = 10 
print(a > 18)
```

```python
# if elif else 

marks = int(input("Enter your marks : "))

if marks >= 90:
    print("A grade")
elif marks >= 75:
    print("B grade")
elif marks >= 50:
    print("C grade")
else:
    print("Fail") 
```

```python
# Logical operators 

age = 61

if age >= 18 and age <= 60:
    print("Your are eligible")
else:
    print("ur are not eligible ")
```

```python
# For loop 

for i in range(1, 100):
    print(i * 10)
```

```python
# while loop 

i = 1
while i <= 10:
    print(i)
    i += 1
```
