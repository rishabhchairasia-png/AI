```python
# Problem 1: Even or odd

a = int(input("enter a number :"))

if a % 2 == 0:
    print("even")
else:
    print("odd")
```

```python
# Problem 2: Positive or negative

n = int(input("enter a number :"))
if n > 0:
    print("positive ")
elif n < 0 :
    print("negative ")
else:
    print("zero")
```

```python
# Problem 3: Largest of two numbers 

a = int(input("enter first number :"))
b = int(input("enter second number :"))

if a > b:
    print("a is largest")
else:
    print("b is largest")
```

```python
# Problem 4: Largest of three numbers

a = int(input("enter first number :"))
b = int(input("enter second number :"))
c = int(input("enter third number :"))

if a > b and a > c:
    print("a is largest")
elif b > a and b >c:
    print("b is largest")
else:
    print("c is largest")
```

```python
# Problem 5: leap year or not

year = int(input("enter a year :"))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Leap year")
else:
    print("Not a leap year")
```

```python
# Problem 6: Prime number or not 

n = int(input("enter a number :"))

if n < 2:
    print("not prime number")
else:
    isprime = True
    for i in range(2 , n):
        if n % i == 0:
            isprime = False
            break
    if isprime:
        print("Prime number")
```

```python
# Problem 7: Factorial of a number

n = int(input("enter a number :"))

factorial = 1

for i in range(1,n+1):
    factorial *= i

print(factorial)
```

```python
# Problem 8: Fibonacci series

n = int(input("enter a number :"))
a = 0
b = 1

for i in range(n):
    print(a)
    a, b = b,a+b
```

```python
# Problem 9 : Reverse a nummber 

n = int(input("enter a number :"))
rev = 0 

while n >  0:
    digit = n%10
    rev = rev * 10 + digit 
    n //= 10
print(rev)
```

```python
# Probelm 10: Palindrome or not

n = int(input("enter a number :"))
og = n 
rev = 0  

while n > 0 :
    digit = n %10

    rev = rev * 10 +digit
    n //= 10

if og == rev:
    print("Pallindrome")
else:
    print("Not a pallindrome")
```
